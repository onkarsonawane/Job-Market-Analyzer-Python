```python
import pandas as pd
import mysql.connector

# १. CSV फाईलमधून डेटा लोड करणे
try:
    df = pd.read_csv('jobs.csv')
    print("📊 मूळ एक्सेल डेटा यशस्वीरित्या लोड झाला आहे!")
except FileNotFoundError:
    print("❌ एरर: jobs.csv फाईल सापडली नाही. कोड असलेल्या फोल्डरमध्येच फाईल ठेवा.")
    exit()

# २. Python Pandas वापरून डेटा फिल्टर करणे (फक्त पुणे आणि ३ लाखापेक्षा जास्त पगार)
pune_jobs = df[(df['Location'] == 'Pune') & (df['Salary'] >= 300000)]

print("\n--- फिल्टर केलेला पुणे लोकेशनचा डेटा ---")
print(pune_jobs)

# ३. MySQL Workbench डेटाबेसशी कनेक्ट करणे
try:
    conn = mysql.connector.connect(
        host="localhost",
        user="root",
        password="Onkar7666",  # तुझा पासवर्ड आधीच व्हेरीफाय झाला आहे
        database="JobAnalyzerDB"
    )
    cursor = conn.cursor()
    
    # डेटाबेसमध्ये डेटा इन्सर्ट करणे
    for index, row in pune_jobs.iterrows():
        sql = "INSERT INTO pune_jobs (title, company, skill, salary) VALUES (%s, %s, %s, %s)"
        cursor.execute(sql, (row['Job_Title'], row['Company'], row['Required_Skill'], int(row['Salary'])))
    
    conn.commit()
    print("\n✅ फिल्टर केलेला डेटा यशस्वीरित्या MySQL डेटाबेसमध्ये सेव्ह झाला आहे!")

except mysql.connector.Error as err:
    print(f"\n❌ डेटाबेस एरर: {err}")
finally:
    if 'conn' in locals() and conn.is_connected():
        cursor.close()
        conn.close()
```
